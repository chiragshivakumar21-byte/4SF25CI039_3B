# Task 6 – Data Visualization


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("swiggy.csv")

print(df.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df['Avg ratings'].dropna(), bins=12, color='coral', edgecolor='black')
plt.xlabel('Average Rating')
plt.ylabel('Number of Restaurants')
plt.title('Swiggy Restaurant Rating Distribution')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
city_counts = df['City'].value_counts().head(10)

plt.figure(figsize=(10, 5))
plt.barh(city_counts.index, city_counts.values, color='teal')
plt.xlabel('Restaurant Count')
plt.ylabel('City')
plt.title('Top 10 Cities with Most Restaurants')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
food_counts = df['Food type'].value_counts().head(8)

plt.figure(figsize=(8, 6))
plt.pie(food_counts.values, labels=food_counts.index,
        autopct='%1.1f%%', startangle=90)
plt.title('Popular Food Categories')
plt.show()
```

```python
delivery_data = df.groupby('City')['Delivery time'].mean().nlargest(10)

plt.figure(figsize=(9, 5))
plt.bar(delivery_data.index, delivery_data.values, color='orange')
plt.xlabel('City')
plt.ylabel('Average Delivery Time')
plt.title('Cities with Higher Average Delivery Time')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
delivery_counts = df['Delivery time'].value_counts().sort_index().head(15)

plt.figure(figsize=(9, 5))
plt.bar(delivery_counts.index.astype(str), delivery_counts.values,
        color='steelblue')
plt.xlabel('Delivery Time')
plt.ylabel('Number of Restaurants')
plt.title('Restaurant Distribution by Delivery Time')
plt.tight_layout()
plt.show()
```

```python
price_trend = df.groupby('City')['Price'].mean().head(10)

plt.figure(figsize=(8, 5))
plt.plot(price_trend.index, price_trend.values, marker='o')

plt.xlabel('City')
plt.ylabel('Average Price')
plt.title('Average Restaurant Price by City')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```
