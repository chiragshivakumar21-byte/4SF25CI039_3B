# Task 3 – Data Transformation


```python
import pandas as pd
import numpy as np

df = pd.read_csv("swiggy.csv")

print("Original dataset:")
print(df.head())
```

```python
swiggy = df.copy()
```

```python
print("Original column names:")
print(swiggy.columns.tolist())
```

```python
swiggy = swiggy.drop_duplicates(subset=['Restaurant', 'Area'])

print("Dataset after removing duplicate restaurants:")
print(swiggy.head())
```

```python
swiggy = swiggy.rename(columns={
    'Restaurant': 'restaurant_name',
    'Avg ratings': 'average_rating',
    'Total ratings': 'total_ratings',
    'Food type': 'food_category',
    'Delivery time': 'delivery_time',
    'Price': 'restaurant_price',
    'City': 'city_name'
})
swiggy['food_count'] = swiggy['food_category'].apply(
    lambda x: len(str(x).split(','))
)
print("Restaurant names with food categories:")
print(swiggy[['restaurant_name', 'food_category', 'food_count']].head(10))
```

```python
swiggy['restaurant_price'] = pd.to_numeric(
    swiggy['restaurant_price']
)

swiggy['average_rating'] = pd.to_numeric(
    swiggy['average_rating']
)

swiggy['price_category'] = np.where(
    swiggy['restaurant_price'] > 300, 'Expensive', 'Affordable'
)
print("Price category of restaurants:")
print(swiggy[['restaurant_name', 'restaurant_price', 'price_category']].head(10))
```

```python
print("Missing values after transformation:")
print(swiggy.isnull().sum())
```

```python
print("Updated column names:")
print(swiggy.columns.tolist())
swiggy.to_csv("transformed_data.csv", index=False)
```
