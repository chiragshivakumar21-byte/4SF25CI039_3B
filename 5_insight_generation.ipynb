# Task 5 – Insight Generation


```python
import pandas as pd
import numpy as np

df = pd.read_csv("swiggy.csv")
```

```python
city_data = df['City'].value_counts().head(10).reset_index()
city_data.columns = ['City', 'Number of Restaurants']

city_data
```

```python
rating_data = df.groupby('Restaurant')['Avg ratings'].mean().reset_index()
rating_data = rating_data.sort_values(by='Avg ratings', ascending=False).head(10)

rating_data
```

```python
food_data = df['Food type'].value_counts().head(10).reset_index()
food_data.columns = ['Food Type', 'Restaurant Count']

food_data
```

```python
city_data = df.groupby('City').agg(
    Restaurant_Count=('Restaurant', 'count'),
    Average_Price=('Price', 'mean'),
    Average_Rating=('Avg ratings', 'mean'),
    Total_Ratings=('Total ratings', 'sum'),
    Average_Delivery_Time=('Delivery time', 'mean')
).reset_index()

city_data = city_data.sort_values(by='Restaurant_Count', ascending=False).head(10)

city_data
```

```python
df['Price Category'] = pd.cut(
    df['Price'],
    bins=[0, 200, 400, float('inf')],
    labels=['Low Price', 'Medium Price', 'High Price']
)

price_data = df.groupby('Price Category', observed=False).agg(
    Restaurant_Count=('Restaurant', 'count'),
    Average_Price=('Price', 'mean'),
    Average_Rating=('Avg ratings', 'mean'),
    Total_Ratings=('Total ratings', 'sum'),
    Average_Delivery_Time=('Delivery time', 'mean')
).reset_index()

price_data
```

```python
average_price = df['Price'].mean()

price_category = pd.DataFrame({'Category': ['Affordable', 'Expensive'],
'Number of Restaurants': [len(df[df['Price'] < average_price]),len(df[df['Price'] > average_price])]
})

price_category
```
