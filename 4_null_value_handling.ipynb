# Task 4 – Null Value Handling

```python
import pandas as pd
import numpy as np

df = pd.read_csv("swiggy.csv")

df.head()
```

```python
null_values = df.isnull().sum().reset_index()

null_values.columns = ['Column Name', 'Null Count']

null_values
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_data = pd.DataFrame({
    'Column Name': df.columns,
    'Missing Values': df.isnull().sum().values,
    'Missing Percentage': missing_percentage.values
})

missing_data
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

column_means = df[numeric_columns].mean()

column_means
```

```python
df[numeric_columns] = df[numeric_columns].fillna(column_means)

df.head()
```

```python
remaining_nulls = df.isnull().sum().reset_index()

remaining_nulls.columns = ['Column Name', 'Remaining Null Values']

remaining_nulls
```
